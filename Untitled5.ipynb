```python
from unsloth import FastLanguageModel
import torch
```

```python
MODEL="unsloth/gpt-oss-20b"
```

```python
model,tokenizer=FastLanguageModel.from_pretrained(
    model_name=MODEL,
    max_seq_length=2028,
    dtype=None,
    load_in_4bit=True,
    full_finetuning=False
)
```

```python
model=FastLanguageModel.get_peft_model(
    model,
    target_modules=["q_proj","k_proj","v_proj","o_proj"],
    use_gradient_checkpointing="unsloth",
    r=16,
    lora_alpha=4,
    lora_dropout=0,
    bias="none"
)
```

```python
from datasets import load_dataset

name="HuggingFaceH4/Multilingual-Thinking"
reasoning_data=load_dataset(name,split="train")

reasoning_data[0]
```

```python
from unsloth.chat_templates import standardize_sharegpt

def formatting_prompts_func(examples):
  convos = examples["messages"]
  texts=[tokenizer.apply_chat_template(convo,tokenize=False) for convo in convos]
  return {"text":texts}

dataset = standardize_sharegpt(reasoning_data)
dataset = dataset.map(formatting_prompts_func,batched=True)
```

```python
from trl import SFTTrainer,SFTConfig

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset,
    args=SFTConfig(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        max_steps=60,
        learning_rate=2e-4,
        optim="adamw_8bit",
        weight_decay=0.01,
    )
)
```

```python
trainer_stats=trainer.train()
```

```python
FastLanguageModel.for_inference(model)

test_prompt="Solve this step by step: If a train travels 120 km in 2 hours, what is its average speed?"

messages=[{"role":"user","content":test_prompt}]
inputs=tokenizer.apply_chat_template(messages,tokenize=True,add_generation_prompt=True,return_tensors="pt").to("cuda")

outputs=model.generate(input_ids=inputs,max_new_tokens=64,use_cache=True)
print(tokenizer.batch_decode(outputs))
```
